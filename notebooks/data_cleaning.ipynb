```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("ecommerce_sales_dataset.csv")
df = df.replace({np.nan: None})
df.head()
```

```python
df.info()
df.describe()
```

```python
df['Order_Date'] = pd.to_datetime(df['Order_Date'], errors='coerce')
df.drop_duplicates(inplace=True)
df.fillna(0, inplace=True)
```

```python
df['Month'] = df['Order_Date'].dt.month
df['Year'] = df['Order_Date'].dt.year

df['Profit_Margin'] = df.apply(
    lambda x: x['Profit']/x['Sales'] if x['Sales'] != 0 else 0, axis=1
)

df.head()
```

```python
customers = df[['Customer_ID','Customer_Name','Gender','Age','City','State','Region']].drop_duplicates()

for _, row in customers.iterrows():
    cursor.execute("""
    INSERT IGNORE INTO customers 
    (customer_id, customer_name, gender, age, city, state, region)
    VALUES (%s,%s,%s,%s,%s,%s,%s)
    """, tuple(row))

conn.commit()
print("Customers loaded")
```

```python
products = df[['Product_ID','Product_Category','Product_Name','Price']].drop_duplicates()

for _, row in products.iterrows():
    cursor.execute("""
    INSERT IGNORE INTO products
    (product_id, product_category, product_name, price)
    VALUES (%s,%s,%s,%s)
    """, tuple(row))

conn.commit()
print("Products loaded")
```

```python
orders = df[['Order_ID','Order_Date','Customer_ID','Payment_Mode','Delivery_Status']].drop_duplicates()

for _, row in orders.iterrows():
    cursor.execute("""
    INSERT IGNORE INTO orders
    (order_id, order_date, customer_id, payment_mode, delivery_status)
    VALUES (%s,%s,%s,%s,%s)
    """, tuple(row))

conn.commit()
print("✅ Orders loaded")
```
